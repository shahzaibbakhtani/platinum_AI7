```python
label = df.copy()
```

```python
label = pd.get_dummies(label , columns = ["City"] , dtype = int)
print(label)
```

```python
from sklearn.linear_model import LinearRegression 
```

```python
Model = LinearRegression()
X = [[5],[8],[9],[10], [11],[13],[15],[20]]
Y = [30,55,68,75,77,80,85,95]
Model.fit(X,Y)
hours = float(input("Enter number of hours you studied:"))
results = Model.predict([[hours]])[0]
print(f"Based on your given study hours {hours}, you probably would get {round(results,2)} marks") 
```

```python
from sklearn.linear_model import LogisticRegression 
```

```python
Model = LogisticRegression()
X = [[5],[8],[9],[10], [11],[13],[15],[20]]
Y = [0,0,0,0,1,1,1,1]
Model.fit(X,Y)
hours = float(input("Enter the number of hours:"))
results = Model.predict([[hours]])[0]

if results == 1:
    print(f"Based on your number of hours studied {hours}, you would get passed")
else :
    print(f"Based on your number of hours studied {hours} , you would get failed")
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
Model = KNeighborsClassifier(n_neighbors = 3)
X = [[180, 6] , [190,7] , [200,8] ,[230,9] , [250,10] , [270 , 11] , [280,13],[290,14], [300,16] ]
Y= [0,0,0,0,1,1,1,1,1]

Model.fit(X,Y)
Weight = float(input("Enter your fruit's weight:"))
Size = float(input("Enter your fruit's size:"))

results = Model.predict([[Weight , Size]])[0]

if results == 0:
    print(f" Based on your given weight {Weight} and size {Size} this is likely an apple")
else :
    print(f"Based on your given weight {Weight} and size {Size} this is likely an orange")
```
