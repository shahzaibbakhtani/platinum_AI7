## Converting First Strring into Date

```python
df["date"] = pd.to_datetime(df["date"] , dayfirst = True)
df.info()
```

```python
df[df["date"] > "2019-12-6"]
```

```python
df["Date_Year"] = df["date"].dt.year
df["Date_Month"] = df["date"].dt.month
df["Day"] = df["date"].dt.day_name()
df["Is Weekend"] = np.where(df["Day"].isin(["Saturday" , "Sunday"]), 1,0)
```

```python
df.head(8)
```

```python
df["Quarter"] = df["date"].dt.quarter
df["Semester"] = np.where(df["Quarter"].isin([1,2]), 1,2) # in this jahan 1,2 hogha us main semeter ki jagha 1 aahe gha 
df
```
