```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
import scipy.stats as stats

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv(r"C:\Users\asifb\OneDrive\Desktop\Python language\Jupyter Notebook\Kaggle\concrete_data.csv")
df.head()
```

```python
# We will find the strength of concrete based on the inputs we have, so linear regression problem 
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
X = df.drop(columns = ["Strength"])
Y = df.iloc[:,-1]
```

```python
X_train , X_test , y_train , y_test =train_test_split(X,Y , test_size = 0.25 , random_state = 42) 
```

```python
# Doing regression without transformation 
Model = LinearRegression()
Model.fit(X_train, y_train)

y_pred = Model.predict(X_test)

print(f"R2-score is {r2_score(y_test , y_pred)}")  #  always y_test then y_pred 
```

```python
for col in X_train.columns:
    plt.figure(figsize = (14,4))
    plt.subplot(1,2,1)
    sns.histplot(X_train[col], kde = True)
    plt.title(col)

    plt.subplot(1,2,2)
    stats.probplot(X_train[col], dist="norm", plot=plt)
    plt.title(col)
    plt.show()
```

```python
from sklearn.preprocessing import PowerTransformer

pt = PowerTransformer(method='box-cox') 

X_trained = pt.fit_transform(X_train+0.000001)
X_tested = pt.transform(X_test + 0.000001)

pd.DataFrame({"columns" : X_train.columns , "lambdas" : pt.lambdas_})
```

```python
Model = LinearRegression()

Model.fit(X_trained , y_train)
y_pred = Model.predict(X_tested)

print(f" R-2 Score is: {r2_score(y_test , y_pred)}")
```

```python
X_trained  = pd.DataFrame(X_trained , columns = X_train.columns)

for col in X_trained.columns :
    plt.figure(figsize = (14,4))
    plt.subplot(1,2,1)
    sns.histplot(X_train[col], kde = True)
    plt.title(f"Before {col}")
    
    plt.subplot(1,2,2)
    sns.histplot(X_trained[col], kde = True)
    plt.title(f"After {col}")

    plt.show()
```
