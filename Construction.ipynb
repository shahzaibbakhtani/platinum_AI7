## Applying Feature Construction

```python
df["Family Size"] = df["SibSp"] + df["Parch"] +1 
df.head()
```

```python
def myfunc(num):
    if num == 1:
        #alone
        return 0
    elif num >1 and num <=4:
        # small family
        return 1
    else:
        # large family
        return 2
```

```python
myfunc(2)
```

```python
df["Family Type"] = df["Family Size"].apply(myfunc)
```

```python
df.sample(5)
```

```python
df.drop(columns = ["SibSp" , "Family Size" , "Parch"] , inplace = True)
df.head()
```

```python
X = df.iloc[:, 1:]
y = df.iloc[:, 0]
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))   # accuracy increased 
```
