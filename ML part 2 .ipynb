```python
import pandas as pd 
import seaborn as sns 
import plotly as px
import matplotlib.pyplot as plt 
```

```python
tips= sns.load_dataset("tips")
```

```python
flights = sns.load_dataset("flights")
```

```python
sns.relplot(data = tips , x ="total_bill" , y = "tip", kind ="scatter", height = 3 ,hue = "sex")
```

```python
gap = px.data.gapminder()
```

```python
HM = gap.pivot(index = "country", columns = "year", values = "lifeExp")
```

```python
sns.heatmap(HM)
```

```python
Asia = gap[gap["continent"] == "Asia"]
```

```python
AHM = Asia.pivot(index = "country", columns = "year" , values = "lifeExp")
```

```python
plt.figure(figsize =(10,12))
sns.heatmap(AHM , annot = True , cmap = "summer", linewidth = 1)
```
