```python
import pandas as pd
import numpy as np

import scipy.stats as stats

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.preprocessing import FunctionTransformer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv(r"C:\Users\asifb\OneDrive\Desktop\Python language\Jupyter Notebook\Kaggle\Titanic-Dataset.csv" ,usecols = ["Age", "Fare", "Survived"])
```

```python
df.isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
X = df.iloc[:,1:3]
Y = df.iloc[:,0]
```

```python
X_train , X_test , y_train , y_test =train_test_split(X,Y , test_size = 0.25 , random_state = 42) 
```

```python
# checking wheather AGE is normally distributed or not 

plt.figure(figsize=(7,3))
plt.subplot(1,2,1)
sns.histplot(X_train["Age"], kde =True)
plt.title("Histogram")

plt.subplot(1,2,2)
stats.probplot(X_train['Age'], dist="norm", plot=plt)
plt.title("QQ-Plot")
plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
sns.histplot(X_train["Fare"] , kde = True)
plt.title("Histogram")

plt.subplot(1,2,2)
stats.probplot(X_train['Fare'], dist="norm", plot=plt)
plt.title("QQ-Plot")
plt.show()

# Fare is right skewed as tail is in right direction 
```

```python
Model1 = LogisticRegression()
Model2 = DecisionTreeClassifier()
```

```python
Model1.fit(X_train,y_train)
Model2.fit(X_train,y_train)
```

```python
LRpre = Model1.predict(X_test)
DTpre = Model2.predict(X_test)

print("Logistic Regression: " , round(accuracy_score(LRpre , y_test)*100,2))
print("Decission Treee :" , round(accuracy_score(LRpre , y_test)*100,2))
```

```python
trf1 = FunctionTransformer(func = np.log1p)
```

```python
X_transformed = trf1.fit_transform(X_train)
X_tested = trf1.fit_transform(X_test)
```

```python
sns.displot(data = X_transformed , x ="Fare" , kind = "hist" , kde = True, height = 5 )
```

```python
trf2 = ColumnTransformer([
("Log" , FunctionTransformer(func = np.log1p) , ["Fare"])], remainder = "passthrough")
```

```python
X_trained2 = trf2.fit_transform(X_train)
X_tested2 = trf2.fit_transform(X_test)
```

```python
Model1.fit(X_trained2, y_train)
Model2.fit(X_trained2 , y_train)
y_pred3 = Model1.predict(X_tested2)
y_pred4 = Model2.predict(X_tested2)

print("Accuracy LR",accuracy_score(y_test,y_pred3))
print("Accuracy DT",accuracy_score(y_test,y_pred4))
```
